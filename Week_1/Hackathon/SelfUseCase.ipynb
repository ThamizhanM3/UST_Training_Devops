1. Show path of file

```python
import os

current_file_path = os.path.abspath('Notes.txt')
print("Current file path:", current_file_path)
```

```python
from pathlib import Path

current_file_path = Path('Notes.txt').resolve()
print("Current file path:", current_file_path)
```

```python
from pathlib import Path

current_dir = Path('Notes.txt').resolve().parent
print("Current directory:", current_dir)
```

```python
import os

current_dir = os.path.dirname(os.path.abspath('Notes.txt'))
print("Current directory:", current_dir)
```

2. Open a txt file and read its contents

```python
with open('Notes.txt', 'r') as file:
   lines = file.readlines()

N = 8
for line in lines[:N]:
   print(line, end='')
```

3. Open a word document and read its contents

```python
from docx import Document
doc = Document("Scenarios for Shell Scripting.docx")

print(doc)
for i in doc.paragraphs:
    print(i.text)
```

4. If there is a word "Create", print that line

```python
for i in doc.paragraphs:
    if 'Create' in i.text:
        print(i.text)
```

5. Zip a group of files

```python
import zipfile

with zipfile.ZipFile("new_file.zip", "w") as zipf:
    zipf.write("Notes.txt")
    zipf.write("Notes.py")
    zipf.write("Scenarios for Shell Scripting.docx")
```

6. Read contents of a zip file

```python
import zipfile

with zipfile.ZipFile("new_file.zip", "r") as zip_ref:
    zip_ref.printdir()
```
