4. Multithreaded file searcher

```python
import os

path = "C:/M3/Programming"
files = [f for f in os.listdir(path) if os.path.isfile(os.path.join(path, f))]
# print(files)

for i in files:
    with open(i, 'r') as file:
        try:
            # if i == 'Hackathon.ipynb':
            #     continue
            lines = file.readlines()
        except Exception:
            pass
    # print(i)
    for x in range(len(lines)):
        if 'Create' in lines[x]:
            print('File name :', i)
            print('Line No :', x)
            print(lines[x].strip())
    print('##################################################')
    print()
    print()
    print('##################################################')
```

14. Disk space

```python
import psutil
ram_info = psutil.virtual_memory()
print(f"Total RAM: {ram_info.total / 1024 / 1024 / 1024:.2f} GB")
print(f"Available RAM: {ram_info.available / 1024 / 1024 / 1024:.2f} GB")
print(f"Used RAM: {ram_info.used / 1024 / 1024 / 1024:.2f} GB")
print(f"RAM Usage Percentage: {ram_info.percent}%")
```

```python
import psutil
processes = psutil.process_iter()
for process in processes:
    print(f"Process ID: {process.pid}, Name: {process.name()}")
```

```python

processes = []
for proc in psutil.process_iter(["pid", "name", "cpu_percent", "memory_percent"]):
    try:
        processes.append(proc.info)
    except (psutil.NoSuchProcess, psutil.AccessDenied):
        continue
    
top5 = sorted(processes, key=lambda p: p["memory_percent"], reverse=True)[:5]


for idx, p in enumerate(top5, 1):
    print(
        f"{idx}. PID: {p['pid']:<6} "
        f"CPU: {p['cpu_percent']:<5}%  "
        f"Memory: {p['memory_percent']:.2f}%  "
        f"Name: {p['name']}"
    )

```
